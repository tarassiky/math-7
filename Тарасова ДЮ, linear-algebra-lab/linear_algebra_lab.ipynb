$$\mathbf X\in\mathbb R^{8\times2},\quad \mathbf y\in\mathbb R^8.$$

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

STUDENT_NAMES = np.array(["Аня","Борис","Вера","Глеб","Дина","Егор","Жанна","Илья"])
X = np.array([[1,2],[2,2],[2,3],[3,4],[4,5],[5,7],[6,8],[7,10]], dtype=float)
y = np.array([48,52,55,63,70,78,85,94], dtype=float)
print("X.shape =", X.shape, "; y.shape =", y.shape)
```

### Задание 1.1. Координаты и длина

**[Бумага]** $\overrightarrow{AB}=B-A=(2,-6,3)$, $|\overrightarrow{AB}|=\sqrt{4+36+9}=7$.

```python
A = np.array([1,2,3], dtype=float)
B = np.array([3,-4,6], dtype=float)
AB = B - A
length = np.linalg.norm(AB)
assert np.allclose(AB, [2,-6,3]) and abs(length-7) < 1e-12
print("AB =", AB, "; |AB| =", length)
```

### Задание 1.2. Угол между векторами

$\mathbf a=(-1,1,0)^{\mathsf T}$, $\mathbf b=(1,-2,2)^{\mathsf T}$.

**[Бумага]** $\mathbf a^{\mathsf T}\mathbf b=-3$; $\cos\varphi=-1/\sqrt2$; $\varphi=135^\circ$.

```python
a = np.array([-1,1,0], dtype=float)
b = np.array([1,-2,2], dtype=float)
dot = a @ b
cos_phi = dot / (np.linalg.norm(a)*np.linalg.norm(b))
phi = np.degrees(np.arccos(np.clip(cos_phi, -1, 1)))
assert np.isclose(dot,-3) and np.isclose(phi,135)
print(f"a·b={dot}, cos φ={cos_phi:.4f}, φ={phi:.2f}°")
```

### Задание 1.3. Ортогональность

**[Бумага]** $3\cdot2+2x-1\cdot2=4+2x=0 \Rightarrow x=-2$.

```python
a = np.array([3,2,-1], dtype=float)
x_value = -2.0
b = np.array([2, x_value, 2], dtype=float)
assert np.isclose(a @ b, 0)
print("x =", x_value, "; a·b =", a @ b)
```

### Задание 1.4. Векторы объектов

**[Бумага]** Илья − Аня $=(6,8)$, $\|\cdot\|=10$.

```python
delta = X[-1] - X[0]
dist = np.linalg.norm(delta)
assert np.allclose(delta, [6,8]) and np.isclose(dist, 10)
print("Илья − Аня =", delta, "; норма =", dist)
```

### Задание 2.3. Произведение матриц

$\mathbf{AB}\ne\mathbf{BA}$ — умножение некоммутативно.

### Задание 4.1. Определитель

$\det= -17\ne0$ → обратима.

```python
A = np.array([[5,2,3],[-4,3,5],[2,3,4]], dtype=float)
det_A = np.linalg.det(A)
assert np.isclose(det_A, -17, atol=1e-9)
print("det A =", det_A)
```

## 5. Базис и координаты

```python
B = np.column_stack([np.array([0,1,2]), np.array([1,0,1]), np.array([-1,2,4])]).astype(float)
rank = np.linalg.matrix_rank(B)
det = np.linalg.det(B)
assert rank == 2 and abs(det) < 1e-9
print("rank =", rank, "; det =", det)
```

## 6. Системы

### Задание 6.1. Квадратная система

```python
A = np.array([[2,-1,1],[3,2,2],[1,-2,1]], dtype=float)
b = np.array([2,-2,1], dtype=float)
x_sol = np.linalg.solve(A, b)
residual = A @ x_sol - b
assert np.allclose(x_sol, [2,-1,-3]) and np.allclose(residual, 0)
print("x =", x_sol, "; невязка =", residual)
```

## 7. Проекция

### Задание 7.1. Проекция на прямую

$\widehat{\mathbf y}=(2,2)^{\mathsf T}$, $\mathbf e=(1,-1)^{\mathsf T}$, $\mathbf a^{\mathsf T}\mathbf e=0$.

```python
yv = np.array([3.0,1.0]); a = np.array([1.0,1.0])
c = (a @ yv) / (a @ a)
y_hat = c * a; e = yv - y_hat
assert np.isclose(c, 2) and np.allclose(y_hat, [2,2]) and np.isclose(a @ e, 0)
print("c =", c, "; y_hat =", y_hat, "; e =", e)

plt.figure(figsize=(5,5))
o = np.zeros(2)
plt.quiver(*o, *yv, angles='xy', scale_units='xy', scale=1, label='y')
plt.quiver(*o, *y_hat, angles='xy', scale_units='xy', scale=1, label='ŷ')
plt.quiver(*y_hat, *e, angles='xy', scale_units='xy', scale=1, label='e')
plt.xlim(-0.5,3.5); plt.ylim(-0.5,3.5); plt.grid(); plt.legend()
plt.gca().set_aspect('equal'); plt.show()
```

## 8. Наименьшие квадраты

### Задание 8.1. Регрессия

```python
D = np.column_stack([np.ones(len(X)), X])
beta, *_ = np.linalg.lstsq(D, y, rcond=None)
y_hat_reg = D @ beta
print("beta =", beta)
print("y_hat =", y_hat_reg)
```

### Задание 8.2. Качество приближения

```python
residuals = y - y_hat_reg
sse = float((residuals**2).sum())
rmse = float(np.sqrt(sse / len(y)))
r2 = 1 - sse / float(((y - y.mean())**2).sum())
print(f"SSE={sse:.4f}, RMSE={rmse:.4f}, R2={r2:.4f}")
assert r2 > 0.99

fig, axes = plt.subplots(1,2, figsize=(11,4))
axes[0].scatter(y, y_hat_reg)
lims = [min(y.min(), y_hat_reg.min())-2, max(y.max(), y_hat_reg.max())+2]
axes[0].plot(lims, lims, 'r--')
axes[0].set_xlabel('наблюдение'); axes[0].set_ylabel('предсказание')
axes[1].bar(STUDENT_NAMES, residuals)
axes[1].axhline(0, color='k', lw=0.5)
axes[1].tick_params(axis='x', rotation=45)
plt.tight_layout(); plt.show()
```

### Задание 8.3. Регрессия как проекция

```python
orth = D.T @ (y - y_hat_reg)
assert np.allclose(orth, 0, atol=1e-9)
print("D^T (y - y_hat) =", orth)
```

## 9. PCA

### Задание 9.1. Главные компоненты

```python
mean_X = X.mean(axis=0); scale_X = X.std(axis=0, ddof=0)
Z = (X - mean_X) / scale_X
C = Z.T @ Z / len(Z)
eigvals, eigvecs = np.linalg.eigh(C)
order = np.argsort(eigvals)[::-1]
eigvals = eigvals[order]; eigvecs = eigvecs[:, order]
ratio = eigvals / eigvals.sum()
scores = Z @ eigvecs[:, :1]
Z_rec = scores @ eigvecs[:, :1].T
X_rec = Z_rec * scale_X + mean_X
print("eigenvalues =", eigvals)
print("explained ratio =", ratio)
print("first component =", eigvecs[:,0])
assert ratio[0] > 0.99

for i in np.argsort(scores[:,0]):
    print(f"  {STUDENT_NAMES[i]:8s}  score = {scores[i,0]:+.4f}")

fig, axes = plt.subplots(1,2, figsize=(11,4.5))
axes[0].scatter(Z[:,0], Z[:,1])
for name, pt in zip(STUDENT_NAMES, Z):
    axes[0].annotate(name, pt, xytext=(4,4), textcoords='offset points', fontsize=8)
line = np.linspace(-2.5,2.5,2)[:,None] * eigvecs[:,0]
axes[0].plot(line[:,0], line[:,1], 'r-', label='PC1')
axes[0].set_title('Стандартизованные данные')
axes[0].axis('equal'); axes[0].legend(); axes[0].grid(alpha=0.3)
axes[1].scatter(X[:,0], X[:,1], label='исходные')
axes[1].scatter(X_rec[:,0], X_rec[:,1], label='восстановленные')
for o_, r_ in zip(X, X_rec):
    axes[1].plot([o_[0], r_[0]], [o_[1], r_[1]], color='0.7', lw=1)
axes[1].set_title('Восстановление по одной компоненте')
axes[1].legend(); axes[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

### Задание 9.2. Польза и ограничения

```python
corr = np.corrcoef(X.T)[0,1]
rmse_rec = np.sqrt(np.mean((X - X_rec)**2))
print("corr(X1,X2) =", corr)
print("RMSE восстановления =", rmse_rec)
```

## 10. SVD

### Задание 10.1. Малоранговое приближение

```python
def make_demo_image(shape=(252,526), seed=0):
    rng = np.random.default_rng(seed)
    h,w = shape
    yy,xx = np.mgrid[0:h,0:w]
    base  = 0.5*np.exp(-((xx-w/3)**2+(yy-h/2)**2)/(2*80**2))
    base += 0.3*np.exp(-((xx-2*w/3)**2+(yy-h/3)**2)/(2*50**2))
    base += 0.15*np.sin(xx/20)*np.cos(yy/15)
    base += 0.05*rng.standard_normal(shape)
    return np.clip(base, 0, 1)

A_img = make_demo_image()
U, s, Vt = np.linalg.svd(A_img, full_matrices=False)
m, n = A_img.shape
errors, storages = [], []
for k in [2, 5, 20, 50]:
    Ak = (U[:, :k] * s[:k]) @ Vt[:k, :]
    err = np.linalg.norm(A_img - Ak, 'fro') / np.linalg.norm(A_img, 'fro')
    errors.append(err); storages.append(k*(m+n+1))
    print(f"k={k:2d}  ошибка={err:.5f}  хранить={k*(m+n+1):6d}  доля={k*(m+n+1)/(m*n):.2%}")
assert all(errors[i] >= errors[i+1]-1e-12 for i in range(len(errors)-1))
```

### Задание 10.2. Связь PCA и SVD

```python
_, _, VtZ = np.linalg.svd(Z, full_matrices=False)
align = abs(float(eigvecs[:,0] @ VtZ[0]))
print("|cos|", align)
assert align > 0.999
```

## 11. Итоговая интерпретация

**Связный текст.** Вектор наблюдения задаёт один объект, матрица данных собирает весь набор строк. Линейное отображение переводит признаки в новое представление, а его матрица — векторы в образ. Ранг показывает число независимых направлений; det сообщает, не схлопнулось ли пространство. Система уравнений восстанавливает неизвестные параметры; если точного решения нет, проекция даёт ближайшую точку образа, а метод наименьших квадратов применяет эту проекцию к $\mathbf y$ и находит коэффициенты. Главная компонента — направление максимальной дисперсии стандартизованных данных; сингулярное число измеряет масштаб направления, а малоранговое приближение оставляет несколько наибольших сингулярных чисел и уменьшает объём с малой ошибкой.

**Регрессия:** вход — $\mathbf D$ и $\mathbf y$, выход — $\beta$ и $\hat{\mathbf y}$; польза — линейное описание, риск — переобучение.

**PCA:** вход — $\mathbf X$, выход — направления и координаты; польза — сжатие и визуализация; риск — игнорируется $\mathbf y$, линейность.

**SVD:** вход — матрица $\mathbf A$, выход — $\mathbf U, \mathbf S, \mathbf V^{\mathsf T}$ и приближение ранга $k$; польза — сжатие; риск — сильная усечка убирает детали.
